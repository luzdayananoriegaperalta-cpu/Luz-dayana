```python
import math

# Ejercicio 1: Capacidad de Silos de Granos
print("--- Calculadora de Silos ---")
radio = float(input("Ingresa el radio del silo (m): "))
altura = float(input("Ingresa la altura del silo (m): "))
densidad_maiz = 0.75
volumen = math.pi * (radio**2) * altura
toneladas = volumen * densidad_maiz
print(f"Resultado: El silo puede almacenar aproximadamente {toneladas:.2f} toneladas.\n")

# Ejercicio 2: Dosificación de Fertilizante
print("--- Dosificación de Fertilizante ---")
ha = float(input("Hectáreas totales: "))
dosis_ha = float(input("Dosis (L/ha): "))
producto_total = ha * dosis_ha
agua_total = producto_total * 500
print(f"Resultado: Necesitas {producto_total}L de producto y {agua_total}L de agua.\n")

# Ejercicio 3: Rendimiento de Pulpa
print("--- Rendimiento de Fruta ---")
peso_total = float(input("Peso total de la fruta (kg): "))
residuos = float(input("Peso de los residuos (kg): "))
pulpa = peso_total - residuos
rendimiento = (pulpa / peso_total) * 100
print(f"Resultado: {pulpa}kg de pulpa ({rendimiento:.1f}% de rendimiento).\n")
```
